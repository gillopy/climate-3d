```python
import netCDF4 as nc
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from scipy.interpolate import griddata

# Ruta del archivo .nc (cambia esto por la ubicación de tu archivo)
file_path = "/content/drive/MyDrive/3c7785ed8c7cf99c2ae80fafa694aef4.nc"

# Cargar el archivo NetCDF4
dataset = nc.Dataset(file_path, mode="r")

# Obtener las variables necesarias
t2m = dataset.variables['t2m'][0, :, :]  # Seleccionar el primer 'valid_time'
lat = dataset.variables['latitude'][:]
lon = dataset.variables['longitude'][:]

# Convertir a Celsius
t2m_celsius = t2m - 273.15

# Crear la figura y el mapa
fig, ax = plt.subplots(figsize=(20, 10), subplot_kw={'projection': ccrs.PlateCarree()})
ax.set_extent([-65, -50, -30, -15], crs=ccrs.PlateCarree())  # Focalizar en Paraguay y alrededores

# Agregar características al mapa
ax.add_feature(cfeature.BORDERS, linestyle=':')
ax.add_feature(cfeature.COASTLINE, linewidth=0.5)
ax.add_feature(cfeature.LAND, edgecolor='black', facecolor='lightgray')
ax.add_feature(cfeature.LAKES, edgecolor='black', facecolor='none')

# Dibujar la temperatura
c = ax.pcolormesh(lon, lat, t2m_celsius, cmap='coolwarm', transform=ccrs.PlateCarree())

# Agregar barra de color
cbar = plt.colorbar(c, orientation='horizontal', pad=0.05)
cbar.set_label('2m Temperature (°C)')

# Lista de ciudades con coordenadas (lat, lon)
ciudades = {
    "Asunción": (-25.2637, -57.5759),
    "Ciudad del Este": (-25.5167, -54.6167),
    "Encarnación": (-27.3306, -55.8667),
    "Villarrica": (-25.75, -56.4333),
    "Pedro Juan Caballero": (-22.555, -55.7333)
}

# Interpolar temperatura en las coordenadas de las ciudades
lon_grid, lat_grid = np.meshgrid(lon, lat)
temperaturas = griddata((lon_grid.ravel(), lat_grid.ravel()), t2m_celsius.ravel(),
                        [(ciudades[ciudad][1], ciudades[ciudad][0]) for ciudad in ciudades], method='linear')

# Agregar etiquetas con la temperatura en cada ciudad
for (ciudad, (lat_c, lon_c)), temp in zip(ciudades.items(), temperaturas):
    ax.text(lon_c, lat_c, f"{ciudad}\n{temp:.1f}°C", fontsize=12, ha='center', va='center',
            bbox=dict(facecolor='white', alpha=0.7, edgecolor='black'))

# Título del gráfico
plt.title("2m Temperature over Paraguay and South America")

# Mostrar el gráfico
plt.show()

# Cerrar el dataset
dataset.close()
```

```python
import netCDF4 as nc
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# Ruta del archivo .nc
file_path = "/content/drive/MyDrive/3c7785ed8c7cf99c2ae80fafa694aef4.nc"

# Cargar el archivo NetCDF4
dataset = nc.Dataset(file_path, mode="r")

# Obtener las variables necesarias
t2m = dataset.variables['t2m'][0, :, :]  # Seleccionar el primer 'valid_time'
lat = dataset.variables['latitude'][:]
lon = dataset.variables['longitude'][:]

# Convertir a Celsius
t2m_celsius = t2m - 273.15

# Crear la figura y el mapa
fig, ax = plt.subplots(figsize=(20, 10), subplot_kw={'projection': ccrs.PlateCarree()})
ax.set_extent([-65, -50, -30, -15], crs=ccrs.PlateCarree())  # Focalizar en Paraguay y alrededores

# Agregar características al mapa
ax.add_feature(cfeature.BORDERS, linestyle=':')
ax.add_feature(cfeature.COASTLINE, linewidth=0.5)
ax.add_feature(cfeature.LAND, edgecolor='black', facecolor='lightgray')
ax.add_feature(cfeature.LAKES, edgecolor='black', facecolor='none')

# Dibujar la temperatura
c = ax.pcolormesh(lon, lat, t2m_celsius, cmap='coolwarm', transform=ccrs.PlateCarree())

# Agregar barra de color
cbar = plt.colorbar(c, orientation='horizontal', pad=0.05)
cbar.set_label('2m Temperature (°C)')

# Lista de ciudades con coordenadas (lat, lon)
ciudades = {
    "Asuncion": (-25.2637, -57.5759),
    "Ciudad del Este": (-25.5167, -54.6167),
    "Encarnación": (-27.3306, -55.8667),
    "Villarrica": (-25.75, -56.4333),
    "Pedro Juan Caballero": (-22.555, -55.7333)
}

# Función para encontrar el índice más cercano en la grilla
def encontrar_indice(valor, array):
    return np.abs(array - valor).argmin()

# Obtener la temperatura más cercana para cada ciudad
for ciudad, (lat_c, lon_c) in ciudades.items():
    idx_lat = encontrar_indice(lat_c, lat)
    idx_lon = encontrar_indice(lon_c, lon)
    temp_ciudad = t2m_celsius[idx_lat, idx_lon]

    # Agregar texto en el mapa
    ax.text(lon_c, lat_c, f"{ciudad}\n{temp_ciudad:.1f}°C", fontsize=12, ha='center', va='center',
            bbox=dict(facecolor='white', alpha=0.7, edgecolor='black'))

# Título del gráfico
plt.title("2m Temperature over Paraguay and South America")

# Mostrar el gráfico
plt.show()

# Cerrar el dataset
dataset.close()
```

```python
import numpy as np
import xarray as xr
import cfgrib
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import matplotlib.animation as animation
import warnings

# Ignorar advertencias
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=UserWarning)

# Cargar el dataset GRIB2
file_path = "gfs.0p25.2025021312.f162.grib2"
data = xr.open_dataset(file_path, engine='cfgrib', filter_by_keys={'typeOfLevel': 'heightAboveGround'})

# Extraer coordenadas y datos de reflectividad (refd)
lons = data.longitude.values
lats = data.latitude.values
height_levels = data.heightAboveGround.values
refd = data.refd.values  # Datos de reflectividad en dB

# Crear la figura y el mapa ortográfico
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection=ccrs.Orthographic(central_longitude=0, central_latitude=0))
ax.set_global()
ax.add_feature(cfeature.LAND, edgecolor='black')
ax.add_feature(cfeature.OCEAN)
ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.BORDERS, linestyle='-')

# Rango de reflectividad
vmin, vmax = np.nanmin(refd), np.nanmax(refd)

# Función de actualización para la animación
def update(frame):
    ax.clear()
    ax.set_global()
    ax.add_feature(cfeature.LAND, edgecolor='black')
    ax.add_feature(cfeature.OCEAN)
    ax.add_feature(cfeature.COASTLINE)
    ax.add_feature(cfeature.BORDERS, linestyle='-')

    # Dibujar la reflectividad
    mesh = ax.pcolormesh(lons, lats, refd[frame], transform=ccrs.PlateCarree(), cmap='turbo', vmin=vmin, vmax=vmax)

    # Agregar título
    ax.set_title(f'Reflectividad Radar - Altura {height_levels[frame]} m')

    return mesh,

# Crear la animación
ani = animation.FuncAnimation(fig, update, frames=len(height_levels), interval=500, blit=False)

# Guardar la animación como video (opcional)
ani.save("animacion_reflectividad.mp4", writer="ffmpeg", fps=5)

plt.show()
```

```python
import netCDF4 as nc
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import QuadMesh
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import geopandas as gpd
from matplotlib.animation import FuncAnimation
from datetime import datetime, timedelta

def create_monthly_animation(dataset_path, geojson_path):
    # Cargar datos
    dataset = nc.Dataset(dataset_path, mode="r")
    gdf = gpd.read_file(geojson_path)

    # Obtener variables
    t2m = dataset.variables['t2m'][:] - 273.15  # Convertir a Celsius
    lat = dataset.variables['latitude'][:]
    lon = dataset.variables['longitude'][:]
    time = dataset.variables['valid_time'][:]

    # Convertir tiempo a fechas
    base_date = datetime(2024, 1, 1)
    dates = [base_date + timedelta(days=int(t)) for t in time]
    months = np.array([d.month for d in dates])

    # Calcular promedios mensuales
    monthly_temps = []
    for month in range(1, 13):
        month_mask = months == month
        if np.any(month_mask):
            monthly_mean = np.mean(t2m[month_mask], axis=0)
            monthly_temps.append(monthly_mean)

    # Configurar el plot con espacio para la barra de color
    fig = plt.figure(figsize=(15, 10))
    ax = fig.add_subplot(1, 1, 1, projection=ccrs.PlateCarree())

    # Establecer los límites para Paraguay y alrededores
    ax.set_extent([-65, -50, -30, -15], crs=ccrs.PlateCarree())

    # Añadir características del mapa
    ax.add_feature(cfeature.BORDERS, linestyle=':')
    ax.add_feature(cfeature.COASTLINE, linewidth=0.5)
    ax.add_feature(cfeature.LAND, edgecolor='black', facecolor='lightgray')
    ax.add_feature(cfeature.LAKES, edgecolor='black', facecolor='none')

    # Añadir límites departamentales
    gdf.boundary.plot(ax=ax, edgecolor="black", linewidth=0.8,
                     transform=ccrs.PlateCarree())

    # Encontrar los valores mínimos y máximos para la escala de color
    vmin = min(np.min(temp) for temp in monthly_temps)
    vmax = max(np.max(temp) for temp in monthly_temps)

    # Crear el primer mesh para poder añadir la barra de color
    mesh = ax.pcolormesh(lon, lat, monthly_temps[0],
                        cmap='coolwarm',
                        transform=ccrs.PlateCarree(),
                        vmin=vmin, vmax=vmax)

    # Función de actualización para la animación
    def update(frame):
        # Eliminar solo los QuadMesh del plot
        for collection in ax.collections:
            if isinstance(collection, QuadMesh):
                collection.remove()

        mesh = ax.pcolormesh(lon, lat, monthly_temps[frame],
                           cmap='coolwarm',
                           transform=ccrs.PlateCarree(),
                           vmin=vmin, vmax=vmax)
        ax.set_title(f'Temperatura Media a 2m - {datetime(2024, frame+1, 1).strftime("%B %Y")}')
        return mesh,

    # Crear la animación
    anim = FuncAnimation(fig, update, frames=len(monthly_temps),
                        interval=1000, blit=True)

    # Añadir barra de color
    fig.colorbar(mesh, ax=ax, orientation='horizontal',
                pad=0.05, label='Temperatura (°C)')

    plt.tight_layout()
    dataset.close()
    return fig, anim

# Para usar la función:
fig, anim = create_monthly_animation(file_path, geojson_path)
plt.show()

# Para guardar la animación (opcional)
# anim.save('temperatura_mensual.gif', writer='pillow')
```

```python
import netCDF4 as nc
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import QuadMesh
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import geopandas as gpd
from matplotlib.animation import FuncAnimation
from datetime import datetime, timedelta

def create_monthly_animation(dataset_path, geojson_path):
    # Cargar datos
    dataset = nc.Dataset(dataset_path, mode="r")
    gdf = gpd.read_file(geojson_path)

    # Obtener variables
    t2m = dataset.variables['t2m'][:] - 273.15  # Convertir a Celsius
    lat = dataset.variables['latitude'][:]
    lon = dataset.variables['longitude'][:]
    time = dataset.variables['valid_time'][:]

    # Convertir tiempo a fechas
    base_date = datetime(2024, 1, 1)
    dates = [base_date + timedelta(days=int(t)) for t in time]
    months = np.array([d.month for d in dates])

    # Calcular promedios mensuales
    monthly_temps = []
    for month in range(1, 13):
        month_mask = months == month
        if np.any(month_mask):
            monthly_mean = np.mean(t2m[month_mask], axis=0)
            monthly_temps.append(monthly_mean)

    # Configurar proyección Ortográfica centrada en Sudamérica
    projection = ccrs.Orthographic(central_longitude=-60, central_latitude=-20)

    fig = plt.figure(figsize=(15, 10))
    ax = fig.add_subplot(1, 1, 1, projection=projection)

    # Añadir características del mapa
    ax.add_feature(cfeature.BORDERS, linestyle=':', linewidth=0.5)
    ax.add_feature(cfeature.COASTLINE, linewidth=0.5)
    ax.add_feature(cfeature.LAND, facecolor='lightgray')
    ax.add_feature(cfeature.OCEAN, facecolor='lightblue')
    ax.add_feature(cfeature.LAKES, facecolor='lightblue')

    # Añadir límites políticos
    gdf.boundary.plot(ax=ax, edgecolor="black", linewidth=0.5,
                     transform=ccrs.PlateCarree())

    # Encontrar los valores mínimos y máximos para la escala de color
    vmin = min(np.min(temp) for temp in monthly_temps)
    vmax = max(np.max(temp) for temp in monthly_temps)

    # Crear el primer mesh
    mesh = ax.pcolormesh(lon, lat, monthly_temps[0],
                        cmap='coolwarm',
                        transform=ccrs.PlateCarree(),
                        vmin=vmin, vmax=vmax)

    # Función de actualización para la animación
    def update(frame):
        for collection in ax.collections:
            if isinstance(collection, QuadMesh):
                collection.remove()

        mesh = ax.pcolormesh(lon, lat, monthly_temps[frame],
                           cmap='coolwarm',
                           transform=ccrs.PlateCarree(),
                           vmin=vmin, vmax=vmax)
        ax.set_title(f'Temperatura Media a 2m - {datetime(2024, frame+1, 1).strftime("%B %Y")}')
        return mesh,

    # Crear la animación
    anim = FuncAnimation(fig, update, frames=len(monthly_temps),
                        interval=1000, blit=True)

    # Añadir barra de color
    fig.colorbar(mesh, ax=ax, orientation='horizontal',
                pad=0.05, label='Temperatura (°C)')

    # Añadir cuadrícula y sombra para efecto 3D
    ax.gridlines(linewidth=0.5, color='gray', linestyle=':')
    ax.stock_img()  # Añade un fondo de relieve básico

    plt.tight_layout()
    dataset.close()
    return fig, anim

# Para usar la función:
fig, anim = create_monthly_animation(file_path, geojson_path)
plt.show()

# Guardar la animación como GIF
anim.save('temperatura_sudamerica.gif', writer='pillow', fps=1, dpi=300)

# O para guardar como video MP4 (requiere ffmpeg)
# anim.save('temperatura_sudamerica.mp4', writer='ffmpeg', fps=1, dpi=100)
```
