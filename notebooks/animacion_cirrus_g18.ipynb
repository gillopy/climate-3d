# Disco completo GOES-West (G18) · reflectancia cirrus C04 (1.4 µm)

Animación del **disco completo** con archivos reales
`OR_ABI-L2-CMIPF-M6C04_G18_*.nc` (NOAA/NESDIS, bucket público `noaa-goes18`).

- **Satélite y banda**: GOES-West (G18), banda **C04 (1.38 µm)**, reflectancia en %.
- **Escala**: percentiles automáticos sobre el primer cuadro (0.00–43.56 %), el mismo
  criterio que `poster_goes.py`; rampa `PALETTES["anomalia"]`, muestreada de la referencia.
- **Cadencia**: 10 min (modo 6 de la banda C04); el nombre del archivo la trae escrita.
- **Vestuario**: el mismo de `poster_goes.py` (`draw_goes_poster` + `caption_lines`),
  así el cuadro fijo y cada cuadro de la animación salen idénticos al póster.

El primer cuadro es **2026-10-03 22:20 UTC** (el mismo `t` de la imagen de referencia:
su escaneo va de 22:20:21 a 22:29:51, y 22:25 es su punto medio).

```python
# ==== configuración ====
from datetime import datetime
from pathlib import Path

# Raíz del proyecto, se corra el notebook desde notebooks/ o desde la raíz.
RAIZ = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

T0       = datetime(2026, 10, 3, 22, 20)   # primer cuadro (UTC) — el de la referencia
N_FRAMES = 12                              # cantidad de cuadros
PASO_MIN = 10                              # cadencia real de C04 en modo 6

SAT, BANDA = "G18", "C04"                  # GOES-West, cirrus 1.38 µm
PALETA     = "anomalia"                    # la rampa muestreada de la referencia
CREDITO    = "Graphic by @tu_usuario"      # tu handle: arriba a la derecha + marca de agua
DESCARGAR  = True                          # False = solo archivos locales; falla si falta un cuadro

# Escala de la barra: None = percentiles automáticos, igual que poster_goes.py.
# Con este cuadro da 0.00 a 43.56 %; la referencia sale de ese mismo criterio.
VMIN, VMAX = None, None
ESCALA_PERCENTILES = (0.5, 99.5)
TOPE_PX = 1200                             # submuestreo del disco, igual que poster_goes.py

LADO_PULG, DPI = 6.0, 120                  # 720 px por cuadro de la animación
CUADRO_MS, FUNDIDO_MS, FUNDIDO_PASOS = 300, 100, 3
```

```python
# ==== datos y utilidades ====
import io, sys
from datetime import timedelta
from pathlib import Path

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image
from IPython.display import Image as MostrarImagen, display
from rich.progress import track

sys.path.insert(0, str(RAIZ / "lib"))
from poster_goes import BG, PALETTES, draw_goes_poster
from goes_fetch import cargar, clave_de, descargar, fundido_desde_imagenes

DATOS = Path(r"D:\Users\guill\dataset era5\goes_c04_g18")
DATOS.mkdir(exist_ok=True)
SALIDAS = Path(r"D:\Users\guill\dataset era5\salidas")
SALIDAS.mkdir(parents=True, exist_ok=True)


def cargar_cuadro(nc):
    """Disco completo submuestreado a TOPE_PX, con los rótulos del póster."""
    return cargar(nc, tope_px=TOPE_PX, escena="FULL DISK")


SLOTS = [T0 + i * timedelta(minutes=PASO_MIN) for i in range(N_FRAMES)]
print(f"{len(SLOTS)} cuadros: {SLOTS[0]:%Y-%m-%d %H:%M} -> {SLOTS[-1]:%Y-%m-%d %H:%M} UTC")
```

```python
# ==== descarga (solo lo que falta) ====
rutas = []
for t in track(SLOTS, description="Descargando cuadros"):
    ruta = descargar(DATOS, SAT, clave_de(SAT, BANDA, t), DESCARGAR)
    rutas.append(ruta)
    print(f"{t:%Y-%m-%d %H:%M} UTC  {ruta.name}  {ruta.stat().st_size / 1e6:.1f} MB")
```

```python
# ==== control: los rótulos deben coincidir con la imagen de referencia ====
campo, x, y, proj, unidad, rotulos = cargar_cuadro(rutas[0])
titulo, subtitulo, fuente, datos = rotulos
if VMIN is None or VMAX is None:
    lo, hi = np.nanpercentile(campo, ESCALA_PERCENTILES)
    VMIN = VMIN if VMIN is not None else float(lo)
    VMAX = VMAX if VMAX is not None else float(hi)
    print(f"escala automática {ESCALA_PERCENTILES}: {VMIN:.2f} a {VMAX:.2f} {unidad}")
print(titulo)
print(subtitulo)
print(fuente)
print(datos)
assert unidad == "%"
assert titulo == "OCTOBER 2026"
assert subtitulo == "G18 GOES-West\nREFLECTANCIA · CIRRUS C04 (1.4 µm)"
assert datos == "DATA: G18 | 2026-10-03T22:25 UTC | MODO Mode 6"
assert fuente == "SOURCE: NOAA/NESDIS G18 ABI L2 CMIP | FULL DISK"
print("\nOK: título, subtítulo, fuente y datos idénticos a la referencia.")
```

```python
# ==== cuadro fijo (PNG) del primer instante ====
fig, ax, _ = draw_goes_poster(
    campo, x=x, y=y, proj=proj, title=titulo, subtitle=subtitulo, unit_label=unidad,
    src_line=fuente, data_line=datos, credit=CREDITO,
    cmap=PALETTES[PALETA], vmin=VMIN, vmax=VMAX)
png = SALIDAS / f"poster_cirrus_{SAT}_{T0:%Y-%m-%dT%H%M}.png"
fig.savefig(png, facecolor=BG)
plt.close(fig)
print("PNG:", png)
display(MostrarImagen(filename=str(png)))
```

```python
# ==== render de todos los cuadros ====
cuadros = []
for i, ruta in enumerate(track(rutas, description="Renderizando"), 1):
    campo_i, x_i, y_i, proj_i, unidad_i, rot_i = cargar_cuadro(ruta)
    fig, ax, _ = draw_goes_poster(
        campo_i, x=x_i, y=y_i, proj=proj_i, title=rot_i[0], subtitle=rot_i[1],
        unit_label=unidad_i, src_line=rot_i[2], data_line=rot_i[3],
        credit=CREDITO, cmap=PALETTES[PALETA], vmin=VMIN, vmax=VMAX,
        size=LADO_PULG, dpi=DPI)
    buffer = io.BytesIO()
    fig.savefig(buffer, format="png", facecolor=BG)
    plt.close(fig)
    cuadros.append(Image.open(buffer).convert("RGB"))
    print(f"  cuadro {i}/{len(rutas)}  {rot_i[3]}")

print(f"{len(cuadros)} cuadros de {cuadros[0].size[0]}x{cuadros[0].size[1]} px")
```

```python
# ==== GIF con fundido entre cuadros ====
fotogramas = fundido_desde_imagenes(cuadros, FUNDIDO_PASOS, CUADRO_MS, FUNDIDO_MS)
gif = SALIDAS / f"animacion_cirrus_{SAT}_{T0:%Y-%m-%dT%H%M}.gif"
fotogramas[0][0].save(
    gif, save_all=True,
    append_images=[im for im, _ in fotogramas[1:]],
    duration=[ms for _, ms in fotogramas], loop=0, optimize=True)
print(f"GIF: {gif}  ({len(fotogramas)} fotogramas, {gif.stat().st_size / 1e6:.1f} MB)")
display(MostrarImagen(filename=str(gif)))
```

## Ideas y variantes

- **Otra ventana**: cambiá `T0` y `N_FRAMES`; la clave S3 se arma sola a partir del
  slot de 10 minutos (no hace falta escribir el nombre del archivo).
- **Otro satélite o banda**: `SAT = "G19"` (GOES-East) o `BANDA = "C13"` (IR 10.3 µm).
  Con temperatura de brillo conviene `PALETA = "termico"` y `VMIN`/`VMAX` en °C.
- **Escala fija**: poné `VMIN`/`VMAX` (por ejemplo `0.0` y `45.0`) para no depender del
  cuadro; si van como `None` se toman los percentiles automáticos del primero.
- **Sin fundido**: `FUNDIDO_PASOS = 0` deja la animación a saltos limpios de 10 min.
- **MP4**: no hace falta ffmpeg en el PATH; el ffmpeg que trae `imageio-ffmpeg`
  ya está disponible (ver la celda de MP4 de `animacion_globo_g19.ipynb`).
- **Crédito**: `CREDITO` alimenta el texto de arriba a la derecha y la marca de agua
  central (esa se dibuja con las letras separadas).
